## 3. Limpieza de Datos

### 3.1. Manejo de Columnas JSON
La columna `player` y `playerCoordinates` contienen datos en formato JSON. Necesitamos extraer la información relevante.

```python
def parse_json_column(df, column_name, keys_to_extract):
    """
    Parsea una columna con strings JSON y extrae claves específicas.
    """
    extracted_data = []
    for item in df[column_name]:
        try:
            # Reemplazar comillas dobles escapadas si es necesario (depende del formato CSV)
            # En este caso, pandas suele manejar el CSV, pero el contenido puede ser un string JSON
            if isinstance(item, str):
                data = json.loads(item)
            else:
                data = {}
            
            extracted_row = {key: data.get(key, np.nan) for key in keys_to_extract}
            extracted_data.append(extracted_row)
        except json.JSONDecodeError:
            extracted_data.append({key: np.nan for key in keys_to_extract})
            
    return pd.DataFrame(extracted_data)

# Extraer datos de 'player'
player_keys = ['name', 'position', 'id']
player_data = parse_json_column(df, 'player', player_keys)
player_data.columns = ['player_name', 'player_position', 'player_id']

# Unir al dataframe original
df = pd.concat([df, player_data], axis=1)

# Mostrar resultado parcial
df[['player', 'player_name', 'player_position']].head()
```

```python
# Extraer datos de 'playerCoordinates'
coords_keys = ['x', 'y']
coords_data = parse_json_column(df, 'playerCoordinates', coords_keys)
coords_data.columns = ['shot_x', 'shot_y']

# Unir al dataframe original
df = pd.concat([df, coords_data], axis=1)

# Eliminar columnas JSON originales para limpiar
df.drop(columns=['player', 'playerCoordinates'], inplace=True)

df.head()
```

### 3.2. Conversión de Tipos de Datos
Aseguramos que `match_date` sea de tipo datetime.

```python
# Convertir match_date a datetime
df['match_date'] = pd.to_datetime(df['match_date'], errors='coerce')

# Verificar conversión
print(df['match_date'].dtype)

# Verificar si hay fechas nulas después de la conversión
missing_dates = df['match_date'].isnull().sum()
print(f"Fechas faltantes: {missing_dates}")
```

### 3.3. Manejo de Valores Nulos
Decidimos qué hacer con los valores nulos encontrados.

```python
# Eliminar filas donde la fecha del partido sea nula (requisito crítico)
df_cleaned = df.dropna(subset=['match_date'])

# Verificar el tamaño del dataset después de eliminar nulos
print(f"Filas originales: {len(df)}")
print(f"Filas después de limpieza: {len(df_cleaned)}")
```

### 3.4. Eliminación de Duplicados
Verificamos si existen registros duplicados.

```python
duplicates = df_cleaned.duplicated().sum()
print(f"Duplicados encontrados: {duplicates}")

if duplicates > 0:
    df_cleaned = df_cleaned.drop_duplicates()
    print("Duplicados eliminados.")
```
