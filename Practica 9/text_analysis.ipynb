# Práctica 9: Análisis de Texto (Word Cloud)

**Objetivo:** Crear nubes de palabras a partir de las variables categóricas de texto en el dataset de la Premier League para visualizar frecuencias de jugadores y situaciones de juego.

```python
import pandas as pd
import matplotlib.pyplot as plt
from wordcloud import WordCloud

# Cargar datos
df = pd.read_csv('cleaned_epl_shots.csv')

# Mostrar las primeras filas de las columnas de texto
text_cols = ['player_name', 'situation', 'bodyPart', 'shotType']
print(df[text_cols].head())
```

## 1. Nube de Palabras: Nombres de Jugadores

Esta nube resaltará a los jugadores que más intentan remates en la liga.

```python
# Concatenar todos los nombres de jugadores en un solo string
players_text = " ".join(df['player_name'].astype(str))

# Generar la nube de palabras
wordcloud_players = WordCloud(width=800, height=400, 
                              background_color='white', 
                              colormap='magma', 
                              collocations=False).generate(players_text)

# Visualizar
plt.figure(figsize=(15, 7.5))
plt.imshow(wordcloud_players, interpolation='bilinear')
plt.axis('off')
plt.title('Nube de Palabras: Jugadores con más Remates', fontsize=20)
plt.show()
```

## 2. Nube de Palabras: Situaciones de Juego y Partes del Cuerpo

Analizaremos las condiciones bajo las cuales ocurren los tiros.

```python
# Combinar columnas de situación y parte del cuerpo
combined_text = " ".join(df['situation'].astype(str)) + " " + " ".join(df['bodyPart'].astype(str))

# Generar nube
wordcloud_combined = WordCloud(width=800, height=400, 
                               background_color='#38003c', # Color Premier League
                               colormap='viridis', 
                               collocations=True).generate(combined_text)

# Visualizar
plt.figure(figsize=(15, 7.5))
plt.imshow(wordcloud_combined, interpolation='bilinear')
plt.axis('off')
plt.title('Nube de Palabras: Situaciones y Partes del Cuerpo', fontsize=20, color='black')
plt.show()
```

## 3. Análisis de Resultados

### Nube de Jugadores:
*   Los nombres más grandes representan a los jugadores que han realizado más disparos en la temporada. 
*   Visualmente, podemos identificar rápidamente a los referentes ofensivos de la liga.

### Nube de Situaciones:
*   Términos como **"assisted"**, **"right-foot"** y **"left-foot"** suelen ser los más prominentes.
*   Esto confirma que la mayoría de los tiros provienen de asistencias y que el pie derecho es la herramienta dominante, aunque hay una presencia significativa de tiros con el pie izquierdo y de situaciones de balón parado (**"set-piece"**).
